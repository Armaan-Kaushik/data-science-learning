# Topic 1: Annotations (`plt.annotate()`)

## Definition

Annotations are used to add explanatory text to specific points on a plot.

Unlike `plt.text()`, which simply places text at given coordinates, `plt.annotate()` can also draw arrows pointing to important data points.

Annotations are widely used in dashboards, research papers, business reports, and presentations to emphasize significant events or values.

```python
import matplotlib.pyplot as plt

months=[1,2,3,4,5]
sales=[20,25,28,55,32]

plt.plot(months,sales)

plt.annotate(
    "Peak Sales",
    xy=(4,55)
)
plt.show()
```

```python
import matplotlib.pyplot as plt

months=[1,2,3,4,5]
sales=[20,25,28,55,32]
plt.plot(months,sales)
plt.annotate(
    "Peak Sales",
    xy=(4,55),
    xytext=(3,65)
)
plt.show()
```

```python
import matplotlib.pyplot as plt

months=[1,2,3,4,5]
sales=[20,25,28,55,32]

plt.plot(months,sales)
plt.annotate(
    "Peak Sales",
    xy=(4,55),
    xytext=(3,65),
    arrowprops=dict(
        arrowstyle="->"
    )
)
plt.show()
```

```python
import matplotlib.pyplot as plt

months=[1,2,3,4,5]
sales=[20,25,28,55,32]
plt.plot(months,sales)
plt.annotate(
    "Peak Sales",
    xy=(4,55),
    xytext=(3,65),
    arrowprops=dict(
        arrowstyle="->",
        color="red"
    )
)
plt.show()
```

```python
import matplotlib.pyplot as plt

fig,ax=plt.subplots()
months=[1,2,3,4,5]
sales=[20,25,28,55,32]
ax.plot(months,sales)
ax.annotate(
    "Peak Sales",
    xy=(4,55),
    xytext=(3,65),
    arrowprops=dict(
        arrowstyle="->"
    )
)
ax.set_title("Monthly Sales")
plt.show()
```

# Topic 2: Advanced Legends

## Definition

A legend explains what different colors, markers, or lines represent in a plot.

Matplotlib allows extensive customization of legends, including:
- Position
- Font size
- Title
- Frame
- Shadow
- Placement outside the plot

Well-designed legends improve readability without hiding important data.

```python
import matplotlib.pyplot as plt

months=[1,2,3,4]

laptop=[20,25,30,28]
mobile=[30,35,40,45]
plt.plot(
    months,
    laptop,
    label="Laptop"
)
plt.plot(
    months,
    mobile,
    label="Mobile"
)
plt.legend()
plt.legend(
    loc="upper left",
    fontsize=12,
    title="Products",
    frameon=False,   #Remove te border 
    shadow=True,
    bbox_to_anchor=(1.05,1),
)
plt.show()
```

# Topic 3: Axis Customization

## Definition

Axes define the coordinate system of a plot.

Matplotlib provides many functions to customize axes, including:

- Axis limits
- Tick positions
- Tick labels
- Axis scaling
- Axis direction
- Tick rotation

Customizing axes improves readability and allows data to be presented more effectively.

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[10,15,12,18,20]
plt.plot(x,y)
plt.xlim(1,4)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[10,15,12,18,20]
plt.plot(x,y)
plt.ylim(0,25)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[10,15,12,18,20]
plt.plot(x,y)
plt.xticks([1,3,5])
plt.show()
```

```python
import matplotlib.pyplot as plt

months=[1,2,3,4]

sales=[20,25,30,28]
plt.plot(months,sales)
plt.xticks(
    [1,2,3,4],
    ["Jan","Feb","Mar","Apr"]
)
plt.show()
```

```python
import matplotlib.pyplot as plt

months=["January","February","March","April"]
sales=[20,25,30,28]

plt.plot(months,sales)
plt.xticks(rotation=45)
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,100,1000,10000]

plt.plot(x,y)
plt.yscale("log")
plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,20,30,40]

plt.plot(x,y)
plt.gca().invert_yaxis()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig,ax=plt.subplots()

x=[1,2,3,4]
y=[10,20,30,40]
ax.plot(x,y)
ax.set_xlim(1,4)
ax.set_ylim(0,50)
ax.set_xticks([1,2,3,4])
ax.set_xticklabels(
    ["Q1","Q2","Q3","Q4"]
)

plt.show()
```

# Topic 4: Multiple Axes (`twinx()` and `twiny()`)

## Definition

Sometimes two datasets share the same x-axis but have completely different scales.

For example:
- Sales (₹)
- Profit (%)

Plotting them on the same y-axis would make one dataset difficult to see.

Matplotlib provides:

- `twinx()` → Creates a second y-axis.
- `twiny()` → Creates a second x-axis.

This allows two different scales to be displayed on the same figure.

```python
import matplotlib.pyplot as plt

months=[1,2,3,4]
sales=[120,150,180,220]
profit=[12,18,16,22]
fig,ax1=plt.subplots()
ax1.plot(
    months,
    sales,
    color="blue",
    label="Sales"
)
ax1.set_ylabel("Sales")
ax2=ax1.twinx()
ax2.plot(
    months,
    profit,
    color="red",
    label="Profit %"
)
ax2.set_ylabel("Profit (%)")
plt.show()
```

# Topic 5: Date & Time Plots

## Definition

A date/time plot is a graph where the x-axis represents dates or times instead of ordinary numbers.

Matplotlib automatically formats date objects on the axis, making it suitable for time-series analysis.

Date plots are commonly used in finance, business analytics, weather forecasting, IoT, and scientific research.

```python
import matplotlib.pyplot as plt
from datetime import datetime
dates=[
    datetime(2024,1,1),
    datetime(2024,1,2),
    datetime(2024,1,3),
    datetime(2024,1,4)
]
sales=[100,120,115,140]
plt.plot(dates,sales)
plt.show()
```

```python
import matplotlib.pyplot as plt
from datetime import datetime
dates=[
    datetime(2024,1,1),
    datetime(2024,1,2),
    datetime(2024,1,3),
    datetime(2024,1,4)
]
sales=[100,120,115,140]
plt.plot(
    dates,
    sales,
    marker="o"
)
plt.show()
```

```python
import matplotlib.pyplot as plt
from datetime import datetime
dates=[
    datetime(2024,1,1),
    datetime(2024,2,1),
    datetime(2024,3,1),
    datetime(2024,4,1)
]
sales=[100,120,150,170]
plt.plot(dates,sales)
plt.gcf().autofmt_xdate()
plt.show()
```

```python
import matplotlib.pyplot as plt
from datetime import datetime
fig,ax=plt.subplots()
dates=[
    datetime(2024,1,1),
    datetime(2024,2,1),
    datetime(2024,3,1),
    datetime(2024,4,1)
]
sales=[100,120,150,170]
ax.plot(dates,sales,marker="o")
fig.autofmt_xdate()
plt.show()
```

# Topic 6: Matplotlib with Pandas

## Definition

Pandas provides built-in plotting methods that use Matplotlib as the visualization engine.

Instead of manually passing lists to Matplotlib, we can directly create plots from DataFrames.

Pandas plotting is convenient for quick analysis, while Matplotlib provides deeper customization.

In fact, `DataFrame.plot()` is a wrapper around Matplotlib.

```python
import pandas as pd
data={
    "Month":["Jan","Feb","Mar","Apr"],
    "Sales":[20,25,30,28]
}
df=pd.DataFrame(data)
print(df)
df.plot(
    x="Month",
    y="Sales"
)

plt.show()
```

```python
df.plot(
    x="Month",
    y="Sales",
    kind="bar"    # Defines which type of plot to use 
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    "Month":["Jan","Feb","Mar","Apr"],
    "Laptop":[20,25,30,28],
    "Mobile":[30,35,40,45]
}
df=pd.DataFrame(data)
df.plot(
    x="Month"
)
plt.show()
```

# Topic 8: DPI (Dots Per Inch)

## Definition

DPI stands for Dots Per Inch.

It determines the resolution (sharpness) of a figure by specifying how many pixels are drawn in every inch of the figure.

Higher DPI produces sharper images but also increases the image size.

DPI is especially important when saving figures for reports, presentations, and printing.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.plot([1,2,3],[10,20,15])
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4),dpi=200)
plt.figure(figsize=(10,6),dpi=80)
plt.plot([1,2,3],[10,20,15])
plt.show()
```

```python
plt.figure(figsize=(4,3),dpi=300)
plt.plot([1,2,3],[10,20,15])
```

# Topic 9: Themes / Styles

## Definition

A style (or theme) is a predefined collection of visual settings that changes the appearance of a plot.

Styles control things like:
- Background color
- Grid appearance
- Font style
- Line colors
- Tick appearance
- Overall layout

Instead of customizing each property manually, we can apply a complete style using `plt.style.use()`.

```python
import matplotlib.pyplot as plt

plt.style.use("ggplot")
x=[1,2,3,4]
y=[10,20,15,25]
plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.style.use("dark_background")
plt.plot([1,2,3],[2,5,4])
plt.show()
```

```python
plt.style.use("fivethirtyeight")
plt.plot([1,2,3],[2,5,4])
plt.show()
```

# Topic 10: Background Customization

## Definition

Background customization changes the color or transparency of a figure and its plotting area.

Matplotlib has two different backgrounds:

1. Figure Background
2. Axes Background

These can be customized independently to improve visualization, presentations, dashboards, or reports.

```python
import matplotlib.pyplot as plt
plt.style.use("default")
plt.plot([1,2,3],[2,5,4])
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(facecolor="lightblue")
plt.plot([1,2,3],[2,5,4])
plt.show()
```

```python
import matplotlib.pyplot as plt
fig=plt.figure(facecolor="lightgrey")
ax=plt.axes()
ax.set_facecolor("lightyellow")
plt.plot([1,2,3],[2,5,4])
plt.show()
```
