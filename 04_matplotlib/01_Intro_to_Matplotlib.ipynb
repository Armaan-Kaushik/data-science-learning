# Topic 2: What is Matplotlib?

## Definition

**Matplotlib** is an open-source Python library used to create static, animated, and interactive visualizations. It provides a wide range of charts and plots that help us understand, analyze, and communicate data effectively.

Matplotlib is the foundation of data visualization in Python and is widely used in Data Analysis, Exploratory Data Analysis (EDA), Machine Learning, Scientific Computing, and Research.

## Import Syntax

```python
import matplotlib.pyplot as plt
```

- `matplotlib` → The main visualization library.
- `pyplot` → A module that provides functions to create and customize plots.
- `plt` → The standard alias used by Python developers.

````python
import matplotlib.pyplot as plt
print("Imported Matplotlib Successfully")
````

````python
print(type(plt))
````

## Creating Your First Plot

## Definition

A **plot** is a graphical representation of data. In Matplotlib, the most basic plot is the **line plot**, which displays data points connected by straight lines.

The `plot()` function is used to create a line plot, while the `show()` function displays the figure.

## Syntax

```python
import matplotlib.pyplot as plt

plt.plot(data)
plt.show()
```

or

```python
import matplotlib.pyplot as plt

x = [x1, x2, x3, ...]
y = [y1, y2, y3, ...]

plt.plot(x, y)
plt.show()
```

````python
import matplotlib.pyplot as plt
numbers=[2,4,6,8,10]
plt.plot(numbers)
plt.show()
````

If you provide only one list, Matplotlib treats it as the y-values and automatically generates the x-values.

````python
months=[1,2,3,4,5]
sales=[120,150,170,160,210]
plt.plot(months,sales)
plt.show()
````

````python
import matplotlib.pyplot as plt

tests = [1, 2, 3, 4]
marks = [65, 72, 78, 90]
plt.plot(tests, marks)
plt.show()
````

````python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
visitors = [200, 240, 260, 250, 310, 380, 420]
plt.plot(days, visitors)
plt.show()
````

````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [120, 150, 170, 160, 210]
plt.plot(months, sales, marker='o')
plt.plot(sales, months, marker='s')

plt.show()
````

# What is a Figure?

## Definition

A **Figure** is the **entire canvas or container** that holds everything you see in a Matplotlib visualization.

It is the top-level object in Matplotlib and can contain one or more **Axes** (plots).

Think of a Figure as a blank sheet of paper. Nothing is drawn on it until you add one or more Axes.

## Syntax

```python
import matplotlib.pyplot as plt

fig = plt.figure()
```

or

```python
fig, ax = plt.subplots()
```

> In future topics, you'll see that `plt.subplots()` creates both a Figure and an Axes together, which is the recommended modern approach.

````python
import matplotlib.pyplot as plt
fig=plt.figure()
plt.show()
````

````python
fig=plt.figure()
plt.plot([10,20,30])
plt.show()
````

````python
import matplotlib.pyplot as plt
plt.plot([10,20,30])
plt.plot([5,15,25])
plt.show()
````

````python
import matplotlib.pyplot as plt

plt.figure()
plt.plot([10,20,30])
plt.figure()
plt.plot([5,15,25])
plt.show()
````

````python
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
plt.show()
````

````python
fig,ax = plt.subplots()
ax.plot([10,20,30])
plt.show()
````

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot([10, 20, 30])
ax.plot([5, 15, 25])
plt.show()
````

````python
fig, ax = plt.subplots(1, 2)
````

# Relationship Between Figure and Axes

## Definition

A **Figure** is the top-level container (canvas) that holds everything in a Matplotlib visualization.

An **Axes** is a plotting area inside the Figure where the actual graph is drawn.

A Figure can contain **one or more Axes**, and every Axes belongs to exactly one Figure.

## Relationship

```
Figure
│
├── Axes
│      ├── X-axis
│      ├── Y-axis
│      ├── Plot
│      ├── Grid
│      ├── Labels
│      └── Title
│
└── Axes
       ├── X-axis
       ├── Y-axis
       └── Plot
```

````python
#One Figure,2 axes
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2)
plt.show()
````

````python
#Two separate Figures
import matplotlib.pyplot as plt

fig1, ax1 = plt.subplots()
fig2, ax2 = plt.subplots()
plt.show()
````

````python
# Adding Multiple lines to one Axes
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
ax.plot([10,20,30])
ax.plot([15,10,25])
plt.show()
````
