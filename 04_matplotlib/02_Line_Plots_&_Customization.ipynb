# Topic 1: Plotting Multiple Lines

## Definition

A **multiple line plot** is a graph that displays two or more datasets on the same Axes using separate lines.

It allows us to compare trends, patterns, and relationships between different variables over the same x-values.

This is one of the most commonly used visualizations in Data Science, Business Analytics, Finance, and Machine Learning.

## Syntax

```python
plt.plot(x, y1)
plt.plot(x, y2)
plt.show()
```

or using the Object-Oriented API

```python
fig, ax = plt.subplots()

ax.plot(x, y1)
ax.plot(x, y2)

plt.show()
```

````python
import matplotlib.pyplot as plt
months=[1,2,3,4,5]
product_A=[100,120,140,160,180]
product_B=[90,110,150,170,200]
plt.plot(months,product_A)
plt.plot(months,product_B)
plt.show()
````

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
A = [10,15,20,18,25]
B = [8,12,18,21,24]
C = [5,9,15,19,22]

plt.plot(months, A)
plt.plot(months, B)
plt.plot(months, C)
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 6)
y1 = x * 2
y2 = x ** 2
plt.plot(x, y1)
plt.plot(x, y2)
plt.show()
````

````python
# Using the Object-Oriented API
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,190,220]
profit = [30,40,45,60,70]
fig, ax = plt.subplots()
ax.plot(months, sales)
ax.plot(months, profit)
plt.show()
````

# Topic 2: Colors in Line Plots

## Definition

Colors are used to distinguish different datasets and improve the readability of a visualization.

Matplotlib supports several ways to specify colors, including:

- Named colors
- Short color codes
- Hexadecimal color codes
- RGB tuples
- RGBA tuples

Choosing appropriate colors helps communicate information clearly and makes visualizations easier to interpret.

## Syntax

```python
plt.plot(x, y, color="red")
```

or

```python
plt.plot(x, y, c="red")
```

Both `color` and `c` are valid.

````python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
sales = [10,15,20,18,25]
profit = [8,10,15,17,22]
plt.plot(x, sales, color="blue")
plt.plot(x, profit, color="red")
plt.show()
````

````python
#Short Color Codes
plt.plot(x, sales, color="b")
plt.plot(x, profit, color="r")
plt.show()
````

````python
#Hexadecimal Colors
plt.plot(x, sales, color="#1F77B4")
plt.plot(x, profit, color="#FF5733")

plt.show()
````

# Topic 3: Line Styles

## Definition

A **line style** determines the visual appearance of a line in a plot. It controls whether the line is solid, dashed, dotted, dash-dot, or follows a custom pattern.

Line styles improve readability, especially when comparing multiple datasets or when graphs are printed in black and white.

## Syntax

```python
plt.plot(x, y, linestyle="-")
```

or

```python
plt.plot(x, y, ls="-")
```

Both `linestyle` and `ls` are valid.

````python
#Solid Line(Default)
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]
plt.plot(x, y, linestyle="-")
plt.show()
````

````python
# Dashed Line
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]
plt.plot(x, y, linestyle="--")
plt.show()
````

````python
# Dotted Line
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]
plt.plot(x, y, linestyle=":")
plt.show()
````

````python
#Dash-Dot
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]
plt.plot(x, y, linestyle="-.")
plt.show()
````

````python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

sales = [100,120,150,170,200]
profit = [20,30,40,50,60]
expenses = [80,90,110,120,140]

plt.plot(x, sales, linestyle="-")
plt.plot(x, profit, linestyle="--")
plt.plot(x, expenses, linestyle=":")

plt.show()
````

# Topic 4: Line Width

## Definition

**Line width** controls the thickness of a plotted line.

By adjusting the line width, we can make important data stand out or reduce the visual emphasis of less important data.

Matplotlib provides the `linewidth` parameter (or its short form `lw`) to control the thickness of a line.

## Syntax

```python
plt.plot(x, y, linewidth=2)
```

or

```python
plt.plot(x, y, lw=2)
```

Both `linewidth` and `lw` are valid.

````python
# Default Line Width
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]

plt.plot(x, y)

plt.show()
````

````python
#Thick Line
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]

plt.plot(x, y, linewidth=4)

plt.show()
````

````python
#Thin Line
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]

plt.plot(x, y, linewidth=0.8)

plt.show()
````

````python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

sales = [100,120,150,170,200]
target = [90,110,140,160,190]

plt.plot(x, sales, linewidth=4)
plt.plot(x, target, linewidth=1)

plt.show()
````

# Topic 5: Markers

## Definition

A **marker** is a symbol placed at each data point on a plotted line.

Markers help identify the exact location of observations, making plots easier to read, especially when there are only a few data points or when multiple lines overlap.

Matplotlib provides the `marker` parameter to customize the appearance of data points.

## Syntax

```python
plt.plot(x, y, marker="o")
```

````python
# Circle Marker
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]

plt.plot(x, y, marker="o")

plt.show()
````

````python
# Square Marker
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]
plt.plot(x, y, marker="s")
plt.show()
````

````python
# Triangle Marker
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]
plt.plot(x, y, marker="^")
plt.show()
````

````python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,15,20,18,25]
plt.plot(x, y, marker="*")
plt.show()
````

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]

sales = [100,120,150,170,200]

plt.plot(
    months,
    sales,
    color="blue",
    linestyle="--",
    linewidth=2,
    marker="o"
)

plt.show()
````

# Topic 6: Axis Labels (`xlabel` and `ylabel`)

## Definition

Axis labels describe what the values on the **X-axis** and **Y-axis** represent.

- The **X-axis label** explains the horizontal axis.
- The **Y-axis label** explains the vertical axis.

Proper axis labels make a visualization self-explanatory and are considered a best practice in data visualization.

## Syntax

```python
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")
```

or using the Object-Oriented API

```python
ax.set_xlabel("X-axis Label")
ax.set_ylabel("Y-axis Label")
```

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
plt.plot(months, sales)
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
revenue = [2500,3200,4100,3900,4800]
plt.plot(months, revenue)
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.show()
````

````python
import matplotlib.pyplot as plt

epochs = [1,2,3,4,5]
accuracy = [72,80,86,91,94]
plt.plot(epochs, accuracy)
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.show()
````

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
fig, ax = plt.subplots()

ax.plot(months, sales)
ax.set_xlabel("Months")
ax.set_ylabel("Sales")
plt.show()
````

# Topic 7: Legends (`plt.legend()`)

## Definition

A **legend** is a box that identifies each plotted dataset by displaying its corresponding label.

Legends make graphs easy to understand when multiple lines, bars, or other plots are displayed together.

To display a legend:
1. Assign a label to each plot using the `label` parameter.
2. Call `plt.legend()` (or `ax.legend()` in the Object-Oriented API).

## Syntax

```python
plt.plot(x, y1, label="Sales")
plt.plot(x, y2, label="Profit")

plt.legend()
```

or using the Object-Oriented API

```python
ax.plot(x, y1, label="Sales")
ax.plot(x, y2, label="Profit")

ax.legend()
```

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
plt.plot(months, sales, label="Sales")
plt.legend()
plt.show()
````

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
profit = [25,30,40,38,55]

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")
plt.legend()
plt.show()
````

````python
months = [1,2,3,4,5]
sales = [120,150,170,160,210]
plt.plot(months, sales, label="Sales")
plt.legend(loc="upper left")
plt.show()
````

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
fig, ax = plt.subplots()
ax.plot(months, sales, label="Sales")
ax.legend()
plt.show()
````

# Topic 8: Titles (`plt.title()`)

## Definition

A **title** is a descriptive heading displayed at the top of a graph. It summarizes what the visualization represents and provides context to the reader.

A meaningful title helps users quickly understand the purpose of the chart.

## Syntax

```python
plt.title("Chart Title")
```

or using the Object-Oriented API

```python
ax.set_title("Chart Title")
```

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
plt.plot(months, sales)
plt.title("Monthly Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
revenue = [2500,3200,4100,3900,4800]
plt.plot(months, revenue)
plt.title("Monthly Revenue (2025)")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.show()
````

````python
import matplotlib.pyplot as plt

epochs = [1,2,3,4,5]
accuracy = [72,80,86,91,94]
plt.plot(epochs, accuracy)
plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.show()
````

# Topic 9: Grid (`plt.grid()`)

## Definition

A **grid** is a set of horizontal and/or vertical reference lines drawn behind a graph to make it easier to read and compare values.

Grids improve readability by helping users estimate the position of data points relative to the axes.

## Syntax

```python
plt.grid(True)
```

or

```python
plt.grid()
```

or using the Object-Oriented API

```python
ax.grid(True)
```

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
plt.plot(months, sales)
plt.grid()
plt.show()
````

````python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
revenue = [2500,3200,4100,3900,4800]
plt.plot(
    months,
    revenue,
    marker="o"
)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.grid()
plt.show()
````

````python
plt.plot(months, sales)
plt.grid(axis="y")
plt.show()
````

````python
plt.plot(months, sales)

plt.grid(
    linestyle="--"
)
plt.show()
````

````python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
fig, ax = plt.subplots()
ax.plot(months, sales)
ax.grid(True,alpha=0.5)
plt.show()
````

# Topic 10: Saving Figures (`plt.savefig()`)

## Definition

`savefig()` is used to save a Matplotlib figure as an image or document file instead of displaying it only on the screen.

It allows you to export plots for reports, presentations, websites, research papers, and dashboards.

## Syntax

```python
plt.savefig("filename.png")
```

or using the Object-Oriented API

```python
fig.savefig("filename.png")
```

````python
# Save as PNG
import matplotlib.pyplot as plt

months = [1,2,3,4,5]
sales = [120,150,170,160,210]
plt.plot(months, sales)
plt.savefig("sales.png")
plt.show()
````

````python
#Save as PDF
plt.plot(months, sales)
plt.savefig("sales_report.pdf")
plt.show()
````

````python
#High Quality Image
plt.plot(months, sales)
plt.savefig(
    "sales.png",
    dpi=300
)
plt.show()
````

````python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.plot(months, sales)
fig.savefig("sales.png")
plt.show()
````

````python
# Transparent Background
plt.plot(months, sales)
plt.savefig(
    "sales.png",
    transparent=True
)
plt.show()
````
