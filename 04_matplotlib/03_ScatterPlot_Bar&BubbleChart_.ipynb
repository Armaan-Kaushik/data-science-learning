# Topic 1: Scatter Plots (`plt.scatter()`)

## Definition

A **scatter plot** is a graph that displays individual data points on a two-dimensional plane. Each point represents one observation with its X-coordinate and Y-coordinate.

Scatter plots are primarily used to **study the relationship, correlation, distribution, and patterns between two numerical variables**.

Unlike a line plot, scatter plots **do not connect the data points with lines**.

## Syntax

```python
plt.scatter(x, y)
```

or using the Object-Oriented API

```python
ax.scatter(x, y)
```

````python
import matplotlib.pyplot as plt
hours=[2,4,5,7,8]
scores=[55,68,72,88,94]
plt.scatter(hours,scores)
plt.show()
````

````python
import matplotlib.pyplot as plt

hours = [1,2,3,4,5,6,7,8]
marks = [40,50,55,62,70,80,90,95]

plt.scatter(hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
````

````python
import matplotlib.pyplot as plt

advertising = [5,10,15,20,25,30]
sales = [40,55,68,82,95,110]
plt.scatter(advertising, sales)
plt.xlabel("Advertising Budget ($1000)")
plt.ylabel("Sales ($1000)")
plt.title("Advertising vs Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

hours = [2,4,5,7,8]
scores = [55,68,72,88,94]
ax.scatter(hours, scores)
ax.set_xlabel("Study Hours")
ax.set_ylabel("Marks")
ax.set_title("Study Hours vs Marks")
plt.show()
````

# Topic 2: Marker Size (`s`)

## Definition

The **`s` parameter** in `plt.scatter()` controls the **size (area)** of each marker (point) in a scatter plot.

It can be:
- A single value, making all markers the same size.
- A list/array, giving each marker a different size.

## Syntax

```python
plt.scatter(x, y, s=size)
```

or using the Object-Oriented API

```python
ax.scatter(x, y, s=size)
```

````python
# Same Size for Every Marker
import matplotlib.pyplot as plt

hours = [2,4,5,7,8]
marks = [55,68,72,88,94]
plt.scatter(
    hours,
    marks,
    s=100
)
plt.show()
````

````python
# Different Size for Every Marker
import matplotlib.pyplot as plt

hours = [2,4,5,7,8]
marks = [55,68,72,88,94]
sizes = [40,80,120,180,250]
plt.scatter(
    hours,
    marks,
    s=sizes
)
plt.show()
````

````python
import matplotlib.pyplot as plt

budget = [5,10,15,20,25]
sales = [40,55,68,82,95]
customers = [80,120,200,260,350]
plt.scatter(
    budget,
    sales,
    s=customers
)
plt.xlabel("Advertising Budget")
plt.ylabel("Sales")
plt.title("Advertising vs Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt

size = [900,1100,1300,1500,1800]
price = [140,170,220,260,320]
bedrooms = [80,120,160,220,300]
plt.scatter(
    size,
    price,
    s=bedrooms
)
plt.xlabel("House Size")
plt.ylabel("Price")
plt.show()
````

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
hours = [2,4,5,7,8]
marks = [55,68,72,88,94]
ax.scatter(
    hours,
    marks,
    s=150
)
plt.show()
````

# Topic 3: Marker Colors (`c`)

## Definition

The **`c` parameter** in `plt.scatter()` controls the **color of the markers** in a scatter plot.

It can be:
- A single color for all markers.
- A list of colors for individual markers.
- A sequence of numerical values used with a colormap to represent another variable.

## Syntax

```python
plt.scatter(x, y, c=color)
```

or using the Object-Oriented API

```python
ax.scatter(x, y, c=color)
```

````python
# Same Color for All Markers
import matplotlib.pyplot as plt
hours = [2,4,5,7,8]
marks = [55,68,72,88,94]
plt.scatter(
    hours,
    marks,
    c="blue"
)
plt.show()
````

````python
 # Different Color for Each Marker
import matplotlib.pyplot as plt

hours = [2,4,5,7,8]
marks = [55,68,72,88,94]
colors = ["red", "blue","green","orange","purple"]
plt.scatter(
    hours,
    marks,
    c=colors
)
plt.show()
````

````python
import matplotlib.pyplot as plt

sales = [10,20,30,40]
profit = [3,7,9,15]
colors = [
    "red",
    "blue",
    "green",
    "orange"
]
plt.scatter(
    sales,
    profit,
    c=colors
)
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit")
plt.show()
````

````python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,7,8]
temperature = [15,20,25,30,35]
plt.scatter(
    x,
    y,
    c=temperature,
    cmap="viridis"
)

plt.colorbar()
plt.show()
````

# Topic 4: Transparency (`alpha`)

## Definition

The **`alpha` parameter** controls the **transparency (opacity)** of markers in a scatter plot.

Its value ranges from **0 to 1**.

- `alpha = 0` → Completely transparent (invisible)
- `alpha = 1` → Completely opaque (default)

Transparency is especially useful when many points overlap, making dense regions easier to identify.

## Syntax

```python
plt.scatter(x, y, alpha=value)
```

or using the Object-Oriented API

```python
ax.scatter(x, y, alpha=value)
```

````python
# Default Transparency
import matplotlib.pyplot as plt
hours = [2,4,5,7,8]
marks = [55,68,72,88,94]
plt.scatter(hours, marks)
plt.show()
````

````python
# Semi-Transparent Markers
import matplotlib.pyplot as plt

hours = [2,4,5,7,8]
marks = [55,68,72,88,94]
plt.scatter(
    hours,
    marks,
    alpha=0.5
)
plt.show()
````

````python
import matplotlib.pyplot as plt

x = [5,5,5,5,5,5,5]
y = [10,10,10,10,10,10,10]
plt.scatter(
    x,
    y,
    s=150,
    alpha=0.3
)
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
income = np.random.normal(50000, 10000, 500)
spending = np.random.normal(3000, 700, 500)
plt.scatter(
    income,
    spending,
    alpha=0.4
)
plt.xlabel("Income")
plt.ylabel("Annual Spending")
plt.title("Customer Spending Pattern")
plt.show()
````

# Topic 5: Bar Charts (`plt.bar()`)

## Definition

A **bar chart** is a visualization used to **compare values across different categories**.

Each category is represented by a rectangular bar, and the height of the bar represents its corresponding value.

Bar charts are ideal for comparing discrete or categorical data, such as sales by product, revenue by department, or students in different classes.

## Syntax

```python
plt.bar(x, height)
```

or using the Object-Oriented API

```python
ax.bar(x, height)
```

````python
import matplotlib.pyplot as plt
products = ["Laptop", "Mobile", "Tablet", "TV"]
sales = [120,180,90,150]
plt.bar(products, sales)
plt.show()
````

````python
students = ["A", "B", "C", "D"]
marks = [82,91,76,88]
plt.bar(students, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
````

````python
departments = ["Sales","HR","IT","Finance"]
employees = [45,18,30,22]
plt.bar(
    departments,
    employees
)
plt.xlabel("Department")
plt.ylabel("Employees")
plt.title("Employees by Department")
plt.show()
````

````python
products = ["Laptop","Mobile","Tablet","TV"]
sales = [120,180,90,150]
plt.bar(
    products,
    sales,
    color="skyblue"
)
plt.show()
````

````python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
products = ["Laptop","Mobile","Tablet","TV"]
sales = [120,180,90,150]
ax.bar(products, sales)
ax.set_title("Product Sales")
ax.set_xlabel("Products")
ax.set_ylabel("Sales")
plt.show()
````

# Topic 6: Horizontal Bar Charts (`plt.barh()`)

## Definition

A **horizontal bar chart** is similar to a vertical bar chart, but the bars extend horizontally instead of vertically.

It is especially useful when category names are long or when comparing ranked data.

## Syntax

```python
plt.barh(y, width)
```

or using the Object-Oriented API

```python
ax.barh(y, width)
```

````python
import matplotlib.pyplot as plt
products = ["Laptop", "Mobile", "Tablet", "TV"]
sales = [120, 180, 90, 150]
plt.barh(products, sales)
plt.show()
````

````python
students = ["Alice", "Bob", "Charlie", "David"]
marks = [88, 92, 75, 85]
plt.barh(students, marks)
plt.xlabel("Marks")
plt.ylabel("Students")
plt.title("Student Marks")
plt.show()
````

````python
departments = [
    "Sales",
    "Human Resources",
    "Information Technology",
    "Finance"
]
employees = [45, 18, 30, 22]
plt.barh(
    departments,
    employees,
    color="teal"
)
plt.xlabel("Employees")
plt.ylabel("Departments")
plt.title("Employees by Department")
plt.show()
````

````python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
products = ["Laptop", "Mobile", "Tablet", "TV"]
sales = [120, 180, 90, 150]
ax.barh(products, sales)
ax.set_title("Product Sales")
ax.set_xlabel("Sales")
ax.set_ylabel("Products")
plt.show()
````

# Topic 7: Grouped Bar Charts

## Definition

A **Grouped Bar Chart** (also called a **Clustered Bar Chart**) is used to compare **multiple groups** across the same categories.

Instead of displaying a single bar for each category, multiple bars are placed **side by side** for easy comparison.

## Syntax

```python
plt.bar(x1, values1, width)
plt.bar(x2, values2, width)
```

or using the Object-Oriented API

```python
ax.bar(x1, values1, width)
ax.bar(x2, values2, width)
```

````python
import matplotlib.pyplot as plt
import numpy as np
products = ["Laptop", "Mobile", "Tablet"]
sales_2024 = [120, 180, 90]
sales_2025 = [150, 210, 110]
x = np.arange(len(products))
width = 0.35

plt.bar(
    x - width/2,
    sales_2024,
    width=width,
    label="2024"
)

plt.bar(
    x + width/2,
    sales_2025,
    width=width,
    label="2025"
)
plt.xticks(x, products)
plt.legend()
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
students = ["A", "B", "C", "D"]
maths = [80, 70, 92, 85]
science = [90, 75, 88, 91]
x = np.arange(len(students))
width = 0.3
plt.bar(
    x-width/2,
    maths,
    width,
    label="Maths"
)
plt.bar(
    x+width/2,
    science,
    width=width,
    label="Science"
)
plt.xticks(x, students)
plt.legend()
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
products = ["Laptop","Mobile","Tablet","TV"]
north = [120,180,90,140]
south = [110,170,100,150]
x = np.arange(len(products))
width = 0.35
plt.bar(
    x-width/2,
    north,
    width,
    label="North"
)
plt.bar(
    x+width/2,
    south,
    width,
    label="South"
)
plt.xticks(x, products)
plt.legend()
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales by Region")
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
products = ["Laptop", "Mobile", "Tablet"]
north = [120,180,90]
south = [110,170,100]
west = [130,175,95]
x = np.arange(len(products))
width = 0.25
plt.bar(x-width, north, width, label="North")
plt.bar(x, south, width, label="South")
plt.bar(x+width, west, width, label="West")
plt.xticks(x, products)
plt.legend()
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
fig, ax = plt.subplots()
products = ["Laptop", "Mobile", "Tablet"]
sales_2024 = [120,180,90]
sales_2025 = [150,210,110]
x = np.arange(len(products))
width = 0.35
ax.bar(x-width/2, sales_2024, width, label="2024")
ax.bar(x+width/2, sales_2025, width, label="2025")
ax.set_xticks(x)
ax.set_xticklabels(products)
ax.legend()
plt.show()
````

# Topic 8: Bubble Charts

## Definition

A **Bubble Chart** is an extension of a scatter plot where the **size of each point represents a third numerical variable**.

It is used to visualize the relationship between **three numerical variables** simultaneously:
- X-axis → First variable
- Y-axis → Second variable
- Bubble Size → Third variable
- (Optional) Bubble Color → Fourth variable

## Syntax

```python
plt.scatter(x, y, s=size)
```

or using the Object-Oriented API

```python
ax.scatter(x, y, s=size)
```

````python
import matplotlib.pyplot as plt
sales = [50, 80, 60, 90]
profit = [10, 18, 12, 22]
customers = [300, 600, 450, 700]
plt.scatter(
    sales,
    profit,
    s=customers
)
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Bubble Chart")
plt.show()
````

````python
import matplotlib.pyplot as plt
sales = [50, 80, 60, 90]
profit = [10, 18, 12, 22]
customers = [300, 600, 450, 700]
bubble_size = [c/2 for c in customers]
plt.scatter(
    sales,
    profit,
    s=bubble_size
)
plt.show()
````

````python
import matplotlib.pyplot as plt
sales = [50,80,60,90]
profit = [10,18,12,22]
customers = [300,600,450,700]
plt.scatter(
    sales,
    profit,
    s=customers,
    c=profit,
    cmap="viridis"
)
plt.colorbar(label="Profit")
plt.show()
````

````python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.scatter(
    sales,
    profit,
    s=customers,
    color="orange"
)
ax.set_title("Bubble Chart")
ax.set_xlabel("Sales")
ax.set_ylabel("Profit")
plt.show()
````
