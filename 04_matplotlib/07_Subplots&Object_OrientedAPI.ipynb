# Topic 1: Introduction to plt.subplot()

## Definition

A subplot allows multiple plots to be displayed inside a single figure.

Instead of creating separate windows for each graph, Matplotlib divides one figure into multiple plotting areas called subplots.

Each subplot behaves like an independent graph with its own title, labels, ticks, and data.

```python
import matplotlib.pyplot as plt
plt.plot([1,2,3],[2,5,4])
plt.title("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.subplot(1,2,1)
plt.plot([1,2,3],[2,5,4])
plt.title("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.subplot(1,2,1)
plt.plot([1,2,3],[2,5,4])
plt.title("Sales")
plt.subplot(1,2,2)
plt.plot([1,2,3],[5,3,6])
plt.title("Profit")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.subplot(2,2,1)
plt.plot([1,2,3],[2,5,4])
plt.title("A")

plt.subplot(2,2,2)
plt.plot([1,2,3],[5,3,6])
plt.title("B")

plt.subplot(2,2,3)
plt.plot([1,2,3],[1,4,2])
plt.title("C")

plt.subplot(2,2,4)
plt.plot([1,2,3],[6,5,7])
plt.title("D")

plt.show()
```

```python
plt.subplot(3,1,1)
plt.subplot(3,1,2)
plt.subplot(3,1,3)
```

```python
plt.subplot(1,3,1)
plt.subplot(1,3,2)
plt.subplot(1,3,3)
```

# Topic 2: plt.subplots()

## Definition

`plt.subplots()` creates a Figure and one or more Axes objects in a single command.

Unlike `plt.subplot()`, which activates one subplot at a time, `plt.subplots()` returns objects that can be directly used to customize each plot.

It is the recommended approach in modern Matplotlib.

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
ax.plot([1,2,3],[2,5,4])
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
ax.plot([1,2,3],[2,5,4])
ax.set_title("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
ax.plot([1,2,3],[2,5,4])
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(1,2)
ax[0].plot([1,2,3],[2,5,4])
ax[0].set_title("Sales")
ax[1].plot([1,2,3],[5,3,6])
ax[1].set_title("Profit")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,2)
ax[0,0].plot([1,2,3],[2,5,4])
ax[0,0].set_title("A")
ax[0,1].plot([1,2,3],[5,3,6])
ax[0,1].set_title("B")
ax[1,0].plot([1,2,3],[3,2,6])
ax[1,0].set_title("C")
ax[1,1].plot([1,2,3],[6,4,5])
ax[1,1].set_title("D")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(8,4))
ax.plot([1,2,3],[2,5,4])
plt.show()
```

# Topic 3: Object-Oriented (OO) API

## Definition

The Object-Oriented (OO) API is the recommended way of creating and customizing plots in Matplotlib.

Instead of using global pyplot functions like `plt.plot()` or `plt.title()`, the OO API works directly with objects such as Figure and Axes.

This approach is more organized, flexible, and suitable for complex visualizations.

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
ax.plot([1,2,3],[2,5,4])
ax.set_title("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots()
fig.suptitle("Company Report")
ax.plot([1,2,3],[2,5,4])
ax.set_title("Sales")
plt.show()
```

# Topic 4: Shared Axes

## Definition

Shared axes allow multiple subplots to use the same x-axis, y-axis, or both.

Instead of repeating identical axis labels and ticks on every subplot, Matplotlib shares them, making figures cleaner and easier to compare.

```python
# Without Shared Axes
import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,1)
ax[0].plot([1,2,3],[10,20,15])
ax[0].set_title("Sales")
ax[1].plot([1,2,3],[5,15,10])
ax[1].set_title("Profit")
plt.show()
```

```python
# Sharex=True
import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,1,sharex=True)
ax[0].plot([1,2,3],[10,20,15])
ax[0].set_title("Sales")
ax[1].plot([1,2,3],[5,15,10])
ax[1].set_title("Profit")
plt.show()
```

```python
#Sharey=True
import matplotlib.pyplot as plt
fig,ax=plt.subplots(1,2,sharey=True)
ax[0].plot([1,2,3],[10,20,15])
ax[0].set_title("Sales")
ax[1].plot([1,2,3],[12,18,14])
ax[1].set_title("Profit")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,2,sharex=True,sharey=True)
ax[0,0].plot([1,2,3],[10,20,15])
ax[0,1].plot([1,2,3],[12,18,14])
ax[1,0].plot([1,2,3],[8,16,11])
ax[1,1].plot([1,2,3],[9,14,13])
plt.show()
```

# Topic 5: GridSpec Introduction

## Definition

GridSpec divides a figure into a flexible grid and allows subplots to occupy one or more grid cells.

Unlike `plt.subplots()`, GridSpec supports unequal subplot sizes and more complex dashboard layouts.

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
fig=plt.figure(figsize=(8,5))
gs=GridSpec(2,2)
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
fig=plt.figure(figsize=(8,5))
gs=GridSpec(2,2)
ax=fig.add_subplot(gs[0,0])
ax.plot([1,2,3],[2,5,4])
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
fig=plt.figure(figsize=(8,5))
gs=GridSpec(2,2)
ax1=fig.add_subplot(gs[0,0])
ax2=fig.add_subplot(gs[0,1])
ax3=fig.add_subplot(gs[1,0])
ax4=fig.add_subplot(gs[1,1])
ax1.set_title("A")
ax2.set_title("B")
ax3.set_title("C")
ax4.set_title("D")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
fig=plt.figure(figsize=(8,5))
gs=GridSpec(2,2)
ax1=fig.add_subplot(gs[:,0])
ax2=fig.add_subplot(gs[0,1])
ax3=fig.add_subplot(gs[1,1])
ax1.set_title("Large Plot")
ax2.set_title("Top Right")
ax3.set_title("Bottom Right")
plt.show()
```

# Topic 6: Complex Figure Layouts

## Definition

Complex figure layouts combine multiple plotting techniques such as `subplots()`, `GridSpec`, shared axes, and the Object-Oriented API to create professional dashboards.

These layouts are commonly used in business reports, data science dashboards, and research papers.

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
fig=plt.figure(figsize=(10,6))
gs=GridSpec(3,2)
ax1=fig.add_subplot(gs[0,:])
ax2=fig.add_subplot(gs[1,0])
ax3=fig.add_subplot(gs[1,1])
ax4=fig.add_subplot(gs[2,0])
ax5=fig.add_subplot(gs[2,1])
ax1.set_title("Sales Trend")
ax2.set_title("Profit")
ax3.set_title("Expenses")
ax4.set_title("Customers")
ax5.set_title("Orders")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
fig=plt.figure(figsize=(10,6))
gs=GridSpec(2,2)
ax1=fig.add_subplot(gs[:,0])
ax2=fig.add_subplot(gs[0,1])
ax3=fig.add_subplot(gs[1,1])
ax1.set_title("Main Chart")
ax2.set_title("Top Right")
ax3.set_title("Bottom Right")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,1,sharex=True,figsize=(8,5))
ax[0].plot([1,2,3],[10,20,15])
ax[1].plot([1,2,3],[8,14,11])
ax[0].set_title("Sales")
ax[1].set_title("Profit")
plt.show()
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(2,2,figsize=(8,6))
fig.suptitle("Company Dashboard")
ax[0,0].set_title("Sales")
ax[0,1].set_title("Profit")
ax[1,0].set_title("Expenses")
ax[1,1].set_title("Customers")
fig.tight_layout()  # automatically adjusts spacing between subplots.
plt.show()
```
