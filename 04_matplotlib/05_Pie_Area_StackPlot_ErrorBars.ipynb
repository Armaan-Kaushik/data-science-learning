# Topic 1: Pie Charts (`plt.pie()`)

## Definition

A **Pie Chart** is a circular chart divided into slices, where each slice represents a proportion of the whole.

Pie charts are useful for showing **part-to-whole relationships**.

For example:

- Market share
- Budget allocation
- Sales contribution
- Population distribution

## Syntax

```python
plt.pie(x)
```

or using the Object-Oriented API

```python
ax.pie(x)
```

````python
import matplotlib.pyplot as plt
sales = [40, 30, 20, 10]
plt.pie(sales)
plt.title("Product Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt
sales = [40, 30, 20, 10]
products = [
    "Laptop",
    "Mobile",
    "Tablet",
    "Watch"
]
plt.pie(
    sales,
    labels=products
)
plt.title("Product Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
sales = [40, 30, 20, 10]
labels = [
    "Laptop",
    "Mobile",
    "Tablet",
    "Watch"
]
ax.pie(
    sales,
    labels=labels
)
ax.set_title("Product Sales")
plt.show()
````

# Topic 2: Pie Chart Customization

## Introduction

Matplotlib provides several parameters to customize pie charts and make them more informative and visually appealing.

The most commonly used parameters are:

- `labels` → Add names to slices
- `colors` → Set slice colors
- `explode` → Separate one or more slices
- `autopct` → Display percentages
- `startangle` → Rotate the chart
- `shadow` → Add a shadow effect

These options are frequently used in business dashboards and reports.

````python
# Labels
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

plt.pie(sales,labels=labels)
plt.title("Product Sales")
plt.show()
````

````python
#Colors
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]
colors=["skyblue","orange","green","red"]

plt.pie(sales,labels=labels,colors=colors)
plt.title("Product Sales")
plt.show()
````

````python
#Explode
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

explode=[0.1,0,0,0]

plt.pie(sales,labels=labels,explode=explode)
plt.title("Highlighting Laptop Sales")
plt.show()
````

````python
 # Show Percentage
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

plt.pie(sales,labels=labels,autopct="%1.1f%%")
plt.title("Product Sales")
plt.show()
````

````python
#Rotate the Pie
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

plt.pie(sales,labels=labels,startangle=90)
plt.title("Rotated Pie Chart")
plt.show()
````

````python
# Shadow
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

plt.pie(sales,labels=labels,shadow=True)
plt.title("Pie Chart with Shadow")
plt.show()
````

````python
 # Detailed  Pie Chart
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]
colors=["#4E79A7","#F28E2B","#59A14F","#E15759"]
explode=[0.08,0,0,0]

plt.pie(
    sales,
    labels=labels,
    colors=colors,
    explode=explode,
    autopct="%1.1f%%",
    startangle=90,
    shadow=True
)

plt.title("Product Sales Distribution")
plt.show()
````

# Topic 3: Donut Charts

## Definition

A **Donut Chart** is a modified version of a pie chart with a hole in the center.

It shows **part-to-whole relationships**, just like a pie chart, but often looks cleaner and leaves space in the middle for additional information such as totals or labels.

Matplotlib does not have a separate `donut()` function. Instead, a donut chart is created by drawing a circle in the center of a pie chart.

````python
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

plt.pie(sales,labels=labels)

centre_circle=plt.Circle((0,0),0.6,color="white")
plt.gca().add_artist(centre_circle)

plt.title("Donut Chart")
plt.show()
````

````python
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

plt.pie(
    sales,
    labels=labels,
    autopct="%1.1f%%"
)

centre_circle=plt.Circle((0,0),0.6,color="white")
plt.gca().add_artist(centre_circle)

plt.title("Product Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]
colors=["#4E79A7","#F28E2B","#59A14F","#E15759"]
explode=[0.08,0,0,0]

plt.pie(
    sales,
    labels=labels,
    colors=colors,
    explode=explode,
    autopct="%1.1f%%",
    startangle=90
)

centre_circle=plt.Circle((0,0),0.6,color="white")
plt.gca().add_artist(centre_circle)

plt.title("Sales Distribution")
plt.show()
````

````python
import matplotlib.pyplot as plt

fig,ax=plt.subplots()

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

ax.pie(
    sales,
    labels=labels,
    autopct="%1.1f%%"
)

centre_circle=plt.Circle((0,0),0.6,color="white")
ax.add_artist(centre_circle)

ax.set_title("Donut Chart")
plt.show()
````

````python
import matplotlib.pyplot as plt

sales=[40,30,20,10]
labels=["Laptop","Mobile","Tablet","Watch"]

plt.pie(
    sales,
    labels=labels,
    autopct="%1.1f%%"
)

centre_circle=plt.Circle((0,0),0.6,color="white")
plt.gca().add_artist(centre_circle)

plt.text(0,0,"Total\n100",ha="center",va="center",fontsize=12)
plt.title("Sales Distribution")

plt.show()
````

# Topic 4: Area Charts (`plt.fill_between()`)

## Definition

An **Area Chart** is a line chart where the area between the line and the x-axis is filled with color.

Area charts emphasize the **magnitude** of values over time, making them useful for showing trends as well as the volume represented by those trends.

## Syntax

```python
plt.fill_between(x, y)
```

or using the Object-Oriented API

```python
ax.fill_between(x, y)
```

````python
import matplotlib.pyplot as plt

months=[1,2,3,4,5]
sales=[20,25,30,28,35]

plt.fill_between(months,sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
````

````python
import matplotlib.pyplot as plt

months=[1,2,3,4,5]
sales=[20,25,30,28,35]

plt.fill_between(
    months,
    sales,
    color="skyblue"
)

plt.title("Monthly Sales")

plt.show()
````

````python
import matplotlib.pyplot as plt

months=[1,2,3,4,5]
sales=[20,25,30,28,35]

plt.fill_between(
    months,
    sales,
    color="green",
    alpha=0.4
)
plt.show()
````

````python
import matplotlib.pyplot as plt

months=[1,2,3,4,5]
sales=[20,25,30,28,35]

plt.plot(months,sales,color="blue")
plt.fill_between(
    months,
    sales,
    color="skyblue",
    alpha=0.4
)
plt.title("Monthly Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt
fig,ax=plt.subplots()

months=[1,2,3,4,5]
sales=[20,25,30,28,35]

ax.plot(months,sales)
ax.fill_between(
    months,
    sales,
    alpha=0.4
)

ax.set_title("Monthly Sales")
plt.show()
````

# Topic 5: Stack Plots (`plt.stackplot()`)

## Definition

A **Stack Plot** (or Stacked Area Chart) is used to visualize how multiple quantities change over time while also showing their cumulative total.

Each dataset is stacked on top of the previous one, making it easy to see both:
- Individual contributions
- Overall total

## Syntax

```python
plt.stackplot(x,y1,y2,...)
```

or using the Object-Oriented API

```python
ax.stackplot(x,y1,y2,...)
```

````python
import matplotlib.pyplot as plt

months=[1,2,3,4]
laptop=[20,25,30,28]
mobile=[30,35,40,45]
tablet=[15,18,20,22]

plt.stackplot(months,laptop,mobile,tablet)
plt.title("Monthly Product Sales")
plt.show()
````

````python
import matplotlib.pyplot as plt

months=[1,2,3,4]
laptop=[20,25,30,28]
mobile=[30,35,40,45]
tablet=[15,18,20,22]

plt.stackplot(
    months,
    laptop,
    mobile,
    tablet,
    labels=["Laptop","Mobile","Tablet"]
)

plt.legend()

plt.show()
````

````python
import matplotlib.pyplot as plt

months=[1,2,3,4]
laptop=[20,25,30,28]
mobile=[30,35,40,45]
tablet=[15,18,20,22]

plt.stackplot(
    months,
    laptop,
    mobile,
    tablet,
    labels=["Laptop","Mobile","Tablet"],
    colors=["skyblue","orange","green"]
)

plt.legend()

plt.show()
````

````python
import matplotlib.pyplot as plt

fig,ax=plt.subplots()

months=[1,2,3,4]
laptop=[20,25,30,28]
mobile=[30,35,40,45]
tablet=[15,18,20,22]

ax.stackplot(
    months,
    laptop,
    mobile,
    tablet,
    labels=["Laptop","Mobile","Tablet"]
)

ax.legend()
ax.set_title("Monthly Sales")

plt.show()
````

# Topic 6: Error Bars (`plt.errorbar()`)

## Definition

An **Error Bar** is a graphical representation of the uncertainty or variability in data.

It shows that a measured value is not exact and may vary within a certain range.

## Syntax

```python
plt.errorbar(x,y,yerr)
```

or

```python
ax.errorbar(x,y,yerr)
```

````python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,15,13,18]
error=[1,2,1,3]

plt.errorbar(x,y,yerr=error)
plt.show()
````

````python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,15,13,18]
error=[1,2,1,3]

plt.errorbar(
    x,
    y,
    yerr=error,
    fmt="o"
)

plt.show()
````

````python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,15,13,18]
error=[1,2,1,3]

plt.errorbar(
    x,
    y,
    yerr=error,
    fmt="o",
    capsize=5
)
plt.show()
````

````python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,15,13,18]
error=[1,2,1,3]

plt.errorbar(
    x,
    y,
    yerr=error,
    fmt="o",
    ecolor="red",
    capsize=5
)

plt.show()
````

````python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,15,13,18]
error=[0.2,0.3,0.1,0.4]

plt.errorbar(
    x,
    y,
    xerr=error,
    fmt="o",
    capsize=5
)

plt.show()
````

````python
import matplotlib.pyplot as plt

fig,ax=plt.subplots()

x=[1,2,3,4]
y=[10,15,13,18]
error=[1,2,1,3]

ax.errorbar(
    x,
    y,
    yerr=error,
    fmt="o",
    capsize=5
)

ax.set_title("Measurements")

plt.show()
````
