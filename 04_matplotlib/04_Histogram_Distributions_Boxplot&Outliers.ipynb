# Topic 1: Histograms (`plt.hist()`)

## Definition

A **Histogram** is a graph used to show the **distribution of numerical data**.

Unlike a bar chart, a histogram groups continuous numerical values into intervals called **bins** and displays how many observations fall into each interval.

## Syntax

```python
plt.hist(x)
```

or using the Object-Oriented API

```python
ax.hist(x)
```

````python
import matplotlib.pyplot as plt
marks = [45, 52, 67, 72, 81, 59, 63, 75, 88, 91]
plt.hist(marks)
plt.show()
````

````python
import matplotlib.pyplot as plt
heights = [
    150, 152, 155, 158, 160,
    162, 165, 167, 170, 172,
    175, 178, 180
]
plt.hist(heights)
plt.title("Distribution of Student Heights")
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
data = np.random.normal(
    loc=50,
    scale=10,
    size=500
)
plt.hist(data)
plt.title("Random Normal Distribution")
plt.show()
````

````python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
marks = [45, 52, 67, 72, 81, 59, 63, 75, 88, 91]
ax.hist(marks)
ax.set_title("Marks Distribution")
ax.set_xlabel("Marks")
ax.set_ylabel("Frequency")
plt.show()
````

# Topic 2: Number of Bins (`bins`)

## Definition

A **bin** is an interval (range) used to group numerical values in a histogram.

The `bins` parameter controls **how many intervals** the data is divided into.

## Syntax

```python
plt.hist(x, bins=10)
```

or using the Object-Oriented API

```python
ax.hist(x, bins=10)
```

````python
import matplotlib.pyplot as plt
marks = [42,45,47,50,54,58,61,66,70,74,78,82]
plt.hist(marks)
plt.title("Default Histogram")
plt.show()
````

````python
import matplotlib.pyplot as plt
marks = [42,45,47,50,54,58,61,66,70,74,78,82]
plt.hist(
    marks,
    bins=5
)
plt.title("Histogram with 5 Bins")
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
data = np.random.normal(
    loc=50,
    scale=10,
    size=500
)
plt.hist(
    data,
    bins=20
)
plt.title("Normal Distribution")
plt.show()
````

#### Topic 3: Histogram Styling

## Definition

Matplotlib provides several parameters to customize the appearance of histograms.

The most commonly used styling parameters are:

- `color` → Changes the fill color of the bars.
- `edgecolor` → Adds borders around the bars.
- `alpha` → Controls transparency.

## Syntax

```python
plt.hist(
    x,
    color="skyblue",
    edgecolor="black",
    alpha=0.7
)
```

or using the Object-Oriented API

```python
ax.hist(
    x,
    color="skyblue",
    edgecolor="black",
    alpha=0.7
)
```

````python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
data = np.random.normal(50, 10, 500)
plt.hist(
    data,
    color="skyblue"
)
plt.title("Histogram with Color")
plt.show()
````

````python
plt.hist(
    data,
    color="skyblue",
    edgecolor="black"
)
plt.title("Histogram with Borders")
plt.show()
````

````python
plt.hist(
    data,
    color="green",
    alpha=0.5
)
plt.title("Transparent Histogram")
plt.show()
````

````python
plt.hist(
    data,
    bins=15,
    color="orange",
    edgecolor="black",
    alpha=0.7
)
plt.title("Styled Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
fig, ax = plt.subplots()
data = np.random.normal(50, 10, 500)
ax.hist(
    data,
    bins=15,
    color="purple",
    edgecolor="black",
    alpha=0.8
)
ax.set_title("Styled Histogram (OO API)")
ax.set_xlabel("Values")
ax.set_ylabel("Frequency")
plt.show()
````

# Topic 4: Understanding Data Distribution

## Definition

A **distribution** describes how the values in a dataset are spread.

A histogram helps us visualize:

- Where most values are concentrated.
- Whether the data is symmetric or skewed.
- Whether there are unusually high or low values.
- The overall shape of the data.

Understanding distributions is a fundamental part of Exploratory Data Analysis (EDA).

````python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)
plt.hist(data, bins=20)
plt.title("Normal Distribution")
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)
data = np.random.exponential(
    scale=10,
    size=1000
)
plt.hist(data, bins=20)
plt.title("Right Skewed Distribution")
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(42)

data = 100 - np.random.exponential(
    scale=10,
    size=1000
)
plt.hist(data, bins=20)
plt.title("Left Skewed Distribution")
plt.show()
````

# Topic 5: Box Plots (`plt.boxplot()`)

## Definition

A **Box Plot** is a statistical visualization that summarizes the distribution of numerical data using the **five-number summary**:

- Minimum
- First Quartile (Q1)
- Median (Q2)
- Third Quartile (Q3)
- Maximum

It also helps identify **outliers** in the dataset.

## Syntax

```python
plt.boxplot(x)
```

or using the Object-Oriented API

```python
ax.boxplot(x)
```

````python
import matplotlib.pyplot as plt
marks = [40,45,50,55,60,65,70,75,80]
plt.boxplot(marks)
plt.title("Student Marks")
plt.show()
````

````python
import matplotlib.pyplot as plt
salary = [
    25000,28000,30000,32000,
    35000,36000,38000,40000,
    42000,45000,50000
]
plt.boxplot(salary)
plt.title("Employee Salaries")
plt.ylabel("Salary")
plt.show()
````

````python
import matplotlib.pyplot as plt
marks = [40,45,50,55,60,65,70,75,80]
plt.boxplot(
    marks,
    vert=False
)
plt.title("Horizontal Box Plot")
plt.show()
````

````python
import matplotlib.pyplot as plt
class_A = [40,50,55,60,65,70,75]
class_B = [35,45,50,60,70,80,90]
plt.boxplot(
    [class_A, class_B]
)
plt.xticks(
    [1,2],
    ["Class A","Class B"]
)
plt.title("Class Comparison")
plt.show()
````

````python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
marks = [40,45,50,55,60,65,70,75,80]
ax.boxplot(marks)
ax.set_title("Marks Distribution")
plt.show()
````

# Topic 6: Quartiles & Interquartile Range (IQR)

## Definition

Quartiles divide a sorted dataset into **four equal parts**.

The three quartiles are:

- **Q1 (First Quartile):** 25% of the data lies below this value.
- **Q2 (Median):** 50% of the data lies below this value.
- **Q3 (Third Quartile):** 75% of the data lies below this value.

The **Interquartile Range (IQR)** measures the spread of the middle 50% of the data.

```text
IQR = Q3 - Q1
```

The IQR is widely used to understand variability and detect outliers.

````python
import numpy as np
marks = [10,20,30,40,50,60,70,80,90]
print(np.percentile(marks, 25))
print(np.percentile(marks, 50))
print(np.percentile(marks, 75))
````

````python
import numpy as np
marks = [10,20,30,40,50,60,70,80,90]
Q1 = np.percentile(marks, 25)
Q3 = np.percentile(marks, 75)
IQR = Q3 - Q1
print(IQR)
````

# Topic 7: Detecting Outliers

## Definition

An **outlier** is a value that is significantly different from the rest of the data.

Outliers may occur because of:

- Data entry errors
- Measurement errors
- Fraudulent transactions
- Genuine rare events

A common statistical method to detect outliers is the **1.5 × IQR Rule**.

## Formula

```text
Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR
```

Any value outside these bounds is considered a potential outlier.

````python
import matplotlib.pyplot as plt
salary = [
25000,28000,30000,
32000,35000,36000,
38000,40000,42000,
45000,50000,
800000
]
plt.boxplot(salary)
plt.title("Salary Distribution")
plt.show()
````

````python
import numpy as np
salary = [
25000,28000,30000,
32000,35000,36000,
38000,40000,42000,
45000,50000,
800000
]
Q1 = np.percentile(salary, 25)
Q3 = np.percentile(salary, 75)
IQR = Q3 - Q1
print(Q1)
print(Q3)
print(IQR)
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print(lower)
print(upper)
````

````python
outliers = []
for value in salary:
    if value < lower or value > upper:
        outliers.append(value)
print(outliers)
````
