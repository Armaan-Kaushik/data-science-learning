Univariate Analysis is the process of analyzing **one variable at a time**.

# Outlier Detection

## What are Outliers?

Outliers are observations that are significantly different from the rest of the data.

These values may represent:

- Data entry errors
- Exceptional business transactions
- Fraudulent activities
- Rare events

Detecting outliers is an important step in Exploratory Data Analysis because they can significantly affect statistical measures such as the mean and standard deviation.

```python
Q1 = df["Revenue"].quantile(0.25)
Q3 = df["Revenue"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df["Revenue"] < lower_bound) |
    (df["Revenue"] > upper_bound)
]
outliers
```

```python
df.loc[0, "Revenue"] = 500000
```

# Bivariate Analysis

## What is Bivariate Analysis?

Bivariate Analysis is the process of studying the relationship between **two variables**.

It helps answer questions like:

- Does revenue vary across regions?
- Which category generates the highest revenue?
- Does increasing quantity increase revenue?
- Are higher discounts associated with higher revenue?

The objective is to identify patterns, trends, and relationships between variables.

# Correlation Analysis

## What is Correlation?

Correlation measures the **strength and direction of the relationship between two numerical variables**.

It helps answer questions like:

- Does increasing quantity increase revenue?
- Does a higher discount lead to higher sales?
- Are price and revenue related?

The correlation coefficient ranges from **-1 to +1**.

- **+1** → Perfect positive correlation
- **0** → No correlation
- **-1** → Perfect negative correlation

# Advanced GroupBy & Aggregation

## Why Use Multiple Aggregations?

In real-world data analysis, managers rarely need just one statistic.

Instead of calculating the total revenue, average revenue, maximum revenue, and order count separately, we can generate all these metrics in a single summary table using the `agg()` function.

This makes reports more efficient, readable, and suitable for business decision-making.

```python
category_summary = (
    df.groupby("Category")
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Average_Revenue=("Revenue", "mean"),
          Maximum_Revenue=("Revenue", "max"),
          Minimum_Revenue=("Revenue", "min"),
          Total_Orders=("Order_ID", "count")
      )
)
category_summary
```

```python
category_summary.sort_values(
    by="Total_Revenue",
    ascending=False
)
```

```python
region_summary = (
    df.groupby("Region")
      .agg(
          Total_Revenue=("Revenue", "sum"),
          Average_Revenue=("Revenue", "mean"),
          Orders=("Order_ID", "count"),
          Average_Discount=("Discount", "mean")
      )
)
region_summary
```

```python
df.groupby(
    ["Region", "Category"]
).agg(
    Revenue=("Revenue", "sum"),
    Orders=("Order_ID", "count")
)
```

# Ranking & Top-N Analysis

## What is Ranking?

Ranking is the process of ordering data from highest to lowest (or vice versa) based on a specific metric.

Businesses often use ranking to identify:

- Top-selling products
- Best customers
- Highest revenue regions
- Best-performing salespeople
- Lowest-performing categories

Pandas provides functions like `nlargest()`, `nsmallest()`, and `sort_values()` to perform these analyses efficiently.

```python
df.nlargest(5, "Revenue")
```

```python
df.nsmallest(5, "Revenue")
```

```python
customer_rank = (
    df.groupby("Customer")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(5)
)
customer_rank
```

```python
category_rank = (
    df.groupby("Category")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)
category_rank
```
